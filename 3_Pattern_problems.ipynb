#### Pattern 1

```python
n = 5 

for i in range(n):
      print("*" * n, end = "\n")   
```

#### Pattern-2: Right-Angled Triangle Pattern

```python
n = 5

for i in range(0,n):
      print("*" * (i + 1))
```

#### Pattern - 3: Right-Angled Number Pyramid

```python
n = 5

nums = []
for i in range(n):
    nums.append(i + 1)
    print(nums)

```

```python
n = 5 

for i in range(n):
    for j in range(i + 1):
        print(j + 1, end = "")
    print() 
```

#### Pattern - 4: Right-Angled Number Pyramid - II

```python
n = 5

for i in range(n):
      for j in range(i+1):
            print(i+1, end = "")
      print()
```

#### Pattern-5: Inverted Right Pyramid

```python
n = 5

for i in range(n, 0, -1):
      print("*" * i)
```

#### Pattern - 6: Inverted Numbered Right Pyramid

```python
n = 5 

for i in range(n,0,-1):
      for j in range(i):
            print(j + 1, end = "")
      print()
```
